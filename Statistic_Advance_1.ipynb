**Question 1**: What is a random variable in probability theory?

**Answer:-** A random variable is a numerical value assigned to the outcomes of a random experiment. It represents the result of
chance events mathematically and is usually denoted by symbols such as X or Y.

**Question 2:** What are the types of random variables?

Answer :-
There are two main types of random variables: (1) Discrete Random Variables – take countable values such as 0, 1,
2, etc. (2) Continuous Random Variables – take any value within a range, such as height, weight, or temperature.

**Question 3:** Explain the difference between discrete and continuous distributions.

**Answer:-**
A discrete distribution deals with countable values and probabilities of specific outcomes, such as the number of
students in a class. A continuous distribution deals with measurements that can take infinite values within a range,
such as time or temperature

**Question 4:** What is a binomial distribution, and how is it used in probability?

Answer :- **bold text**
A binomial distribution is a probability distribution that models the number of successes in a fixed number of
independent trials where each trial has only two outcomes: success or failure. It is widely used in statistics and
probability for experiments like coin tosses and quality testing.

**Question 5:** What is the standard normal distribution, and why is it important?

**Answewr:- **
The standard normal distribution is a normal distribution with mean 0 and standard deviation 1. It is important because
it helps standardize data and is widely used in statistical analysis, hypothesis testing, and probability calculations.

**Question 6:** What is the Central Limit Theorem (CLT), and why is it critical in statistics?

**Answer:-**
The Central Limit Theorem states that the sampling distribution of the sample mean approaches a normal distribution
as the sample size becomes large, regardless of the population distribution. It is important because it allows
statisticians to make inferences about populations using sample data.

**Question 7:** What is the significance of confidence intervals in statistical analysis?

**Answert :-**
Confidence intervals provide a range of values within which the true population parameter is expected to lie with a
certain level of confidence. They help measure the reliability and precision of statistical estimates.

**Question 8:** What is the concept of expected value in a probability distribution?

**Answer:- **
The expected value is the average or mean value of a random variable over many repetitions of an experiment. It is
calculated by multiplying each possible outcome by its probability and summing the results.

**Question 9:** Write a Python program to generate 1000 random numbers from a normal distribution with mean = 50 and standard deviation = 5. Compute its mean and standard deviation using NumPy, and draw a histogram to visualize the distribution.

**Answer :- ** The following Python program generates 1000 random numbers from a normal distribution with mean = 50 and
standard deviation = 5 using NumPy. The computed mean is approximately 50.10 and the standard deviation is
approximately 4.89.

```python
import numpy as np
import matplotlib.pyplot as plt
data = np.random.normal(50, 5, 1000)
mean = np.mean(data)
std = np.std(data)
print("Mean:", mean)
print("Standard Deviation:", std)
plt.hist(data, bins=20)
plt.show()
```

Question 10: You are working as a data analyst for a retail company. The company has collected daily sales data for 2 years and wants you to identify the overall sales trend.

daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
235, 260, 245, 250, 225, 270, 265, 255, 250, 260]

● Explain how you would apply the Central Limit Theorem to estimate the average sales with a 95% confidence interval.

● Write the Python code to compute the mean sales and its confidence interval.

**Answer :-** To estimate the average sales using the Central Limit Theorem (CLT), we calculate the sample mean and use the
t-distribution to create a 95% confidence interval. The sample mean represents the estimated average sales, while the
confidence interval gives the range within which the true average sales is likely to fall. The calculated mean sales is
248.25. The 95% confidence interval is: (240.17, 256.33)

```python
import numpy as np
from scipy import stats
daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255, 235, 260, 245, 250, 225, 270, 265, 255, 250, 260]
mean_sales = np.mean(daily_sales)
std_sales = np.std(daily_sales, ddof=1)
n = len(daily_sales)
confidence = 0.95
t_critical = stats.t.ppf((1 + confidence) / 2, df=n-1)
margin_error = t_critical * (std_sales / np.sqrt(n))
ci_lower = mean_sales - margin_error
ci_upper = mean_sales + margin_error
print("Mean Sales:", mean_sales)
print("95% Confidence Interval:", (ci_lower, ci_upper))
```
